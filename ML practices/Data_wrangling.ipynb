```python
#Rename
df.rename(columns={'Pclass': 'Passenger Class'}).head(2)
```

```python
#old col with names as a key and empty strings as value
col_nam=collections.defaultdict(str)
for name in df.columns:
    col_nam[name]
col_nam
```

```python
#min,amx,mean,count,sum
print('1',df['Age'].max())
print('2',df['Age'].min())
print('3',df['Age'].mean())
#print('4',df['Age'].avg())
print('5',df['Age'].sum())
print('6',df['Age'].count())
```

```python
# grouping rows by time
time_index=pd.date_range('06/06/2017',periods=100000,freq='30S')
```
